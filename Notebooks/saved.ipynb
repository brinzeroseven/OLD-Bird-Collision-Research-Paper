```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataES = {'factor': ['d2edge_IA','d2water_IA','DEM_IA','d2nestsV2_IA','d2streets'],
          'coef score' : [-0.1865,-0.1813,-0.1505,-0.1361,0.0938]
         }
```

```python
dataNVD = {
    'factor': [
        'ASTER ASTGTM2 Elevation',
        'ECMWF Interim Full Daily SFC-FC Soil Water Content L1 0-7 cm',
        'ECMWF Interim Full Daily SFC-FC Albedo',
        'ECMWF Interim Full Daily SFC-FC Water Vapor Concentration',
        'ECMWF Interim Full Daily SFC-FC Total Atmospheric Water'
    ],
    'coef score': [
        0.8175,
        0.6226,
        0.4626,
        0.4631,
        0.4619
    ]
}
```

```python
dataRoadkill = {
    'factor': [
        'Ranger_visits_km',
        'HS_Hyaena_hyaena',
        'MP_Coluber_jugularis',
        'HS_Canis_lupus',
        'HS_Alectoris_chukar'
    ],
    'coef score': [
        0.0777,
        0.0743,
        0.0737,
        0.0728,
        0.0627
    ]
}
```

```python
dfES = pd.DataFrame(dataES)
dfNVD = pd.DataFrame(dataNVD)
dfRoadkill = pd.DataFrame(dataRoadkill)
```

```python
plt.figure(figsize=(8,5))
bars = plt.barh(dfES['factor'],dfES['coef score'])
plt.axvline(0,color='black',linewidth=1)
plt.xlabel('Correlation Coefficient')
plt.ylabel('Factors')
plt.title('Correlation Bar Chart Eagle Subset')
for bar in bars:
    width = bar.get_width()
    text = f'{width:.4f}'
    if abs(width) > 0.05:  # if the bar is big enough, put text inside
        plt.text(width/2, bar.get_y() + bar.get_height()/2, text,
                 ha='center', va='center', color='white', fontsize=9, fontweight='bold')
    else:  # otherwise, put it just outside
        plt.text(width + (0.005 if width > 0 else -0.005), 
                 bar.get_y() + bar.get_height()/2, text,
                 ha='left' if width > 0 else 'right', va='center', color='black', fontsize=9)

plt.tight_layout()
plt.show()
plt.savefig('eaglesubset.png')
```
